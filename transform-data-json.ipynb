## Transforming Data

```python
import pandas as pd

pokemon_info = pd.read_json("../data-files/pokemon_info_generation_1.json")
pokemon_info2 = pd.read_json("../data-files/pokemon_description_generation_1.json")
```

1. Investigate the Data. Check if there is errors, duplicated, etc.

```python
def show_section(title, content):
    print(f"{title}:")
    content()
    print()

sections = [
    ("Shape", lambda: print(pokemon_info.shape)),
    ("DataFrame info", pokemon_info.info),
    ("First five rows", lambda: display(pokemon_info.head())),
    ("Data types", lambda: print(pokemon_info.dtypes)),
    ("Missing values", lambda: print(pokemon_info.isna().sum())),
    ("Duplicate rows", lambda: print(pokemon_info.duplicated().sum())),
    ("Check Impossible Values", lambda: print(pokemon_info.describe()))
]

print("File 1 Infos:")
for title, content in sections:
    show_section(title, content)
```

```python
def show_section(title, content):
    """Print a section title, execute its content, and add a blank line."""
    print(f"{title}:")
    content()
    print()

sections = [
    ("Shape", lambda: print(pokemon_info2.shape)),
    ("DataFrame info", pokemon_info2.info),
    ("First five rows", lambda: display(pokemon_info2.head())),
    ("Data types", lambda: print(pokemon_info2.dtypes)),
    ("Missing values", lambda: print(pokemon_info2.isna().sum())),
    ("Duplicate rows", lambda: print(pokemon_info2.duplicated().sum())),
    ("Check Impossible Values", lambda: print(pokemon_info2.describe()))
]

print("File 2 Infos:")
for title, content in sections:
    show_section(title, content)
```

2. Transform the Data

```python
rename_map = {
    "ID": "pokedex_no",
    "Name": "pokemon",
    "Description": "description",
    "Abilities": "ability",
    "Type": "type",
}

pokemon_info = (
    pokemon_info
    .merge(pokemon_info2[["ID", "Description"]], on="ID", how="left")
    .rename(columns=rename_map)
    [["pokedex_no", "pokemon", "type", "ability", "description"]]
)

display(pokemon_info.head())
```

3. Retrieve the Cleaned Data

```python
pokemon_info.to_json("../data-files/cleaned_pokemon_generation_1.json", index=False)
print("Cleaned data saved successfully.")
```
